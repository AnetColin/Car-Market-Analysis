```python
reference_year = df['Year'].max()

df['Car_Age'] = reference_year - df['Year']

df[['Year', 'Car_Age']].head()
```

```python
df['Depreciation'] = df['Present_Price'] - df['Selling_Price']
```

```python
df['Depreciation_Percentage'] = (
    (df['Present_Price'] - df['Selling_Price'])
    / df['Present_Price']
) * 100
```

```python
df['Car_Name'].unique()
```

```python
# Create Car Age
reference_year = df['Year'].max()
df['Car_Age'] = reference_year - df['Year']
```

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x='Year',
    order=sorted(df['Year'].unique())
)

plt.title('Distribution of Vehicles by Manufacturing Year')
plt.xlabel('Manufacturing Year')
plt.ylabel('Number of Vehicles')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x='Selling_Price',
    bins=30,
    kde=True
)

plt.title('Distribution of Car Selling Prices')
plt.xlabel('Selling Price (Lakhs)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
top_cars = df['Car_Name'].value_counts().head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_cars.values,
    y=top_cars.index
)

plt.title('Top 10 Most Frequently Represented Vehicles')
plt.xlabel('Number of Records')
plt.ylabel('Vehicle')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.countplot(
    data=df,
    x='Fuel_Type'
)

plt.title('Vehicle Distribution by Fuel Type')
plt.xlabel('Fuel Type')
plt.ylabel('Number of Vehicles')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Car_Age',
    y='Selling_Price'
)

plt.title('Selling Price vs Vehicle Age')
plt.xlabel('Car Age (Years)')
plt.ylabel('Selling Price (Lakhs)')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Kms_Driven',
    y='Selling_Price'
)

plt.title('Selling Price vs Kilometers Driven')
plt.xlabel('Kilometers Driven')
plt.ylabel('Selling Price (Lakhs)')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Present_Price',
    y='Selling_Price'
)

plt.title('Present Price vs Selling Price')
plt.xlabel('Present Price (Lakhs)')
plt.ylabel('Selling Price (Lakhs)')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Car_Age',
    y='Depreciation_Percentage'
)

plt.title('Vehicle Age vs Depreciation Percentage')
plt.xlabel('Car Age (Years)')
plt.ylabel('Depreciation (%)')
plt.tight_layout()
plt.show()
```

```python
fuel_dep = (
    df.groupby('Fuel_Type')['Depreciation_Percentage']
    .mean()
    .sort_values()
)

plt.figure(figsize=(8, 6))

sns.barplot(
    x=fuel_dep.index,
    y=fuel_dep.values
)

plt.title('Average Depreciation by Fuel Type')
plt.xlabel('Fuel Type')
plt.ylabel('Average Depreciation (%)')
plt.tight_layout()
plt.show()
```

```python
numeric_cols = [
    'Year',
    'Selling_Price',
    'Present_Price',
    'Kms_Driven',
    'Owner',
    'Car_Age',
    'Depreciation',
    'Depreciation_Percentage'
]

correlation = df[numeric_cols].corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0
)

plt.title('Correlation Matrix of Numerical Variables')
plt.tight_layout()
plt.show()
```
